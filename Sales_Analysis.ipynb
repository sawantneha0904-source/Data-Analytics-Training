DAY 5 - Numpy

NumPy stands for Numerical Python. It is a Python library used to perform numerical calculations efficiently, especially on large amounts of data.


```python
import numpy as np
```

An array is a data structure that stores multiple values in a single variable. In NumPy, arrays are used to store and perform calculations on numerical data efficiently.


Create a 1D array

A 1D (one-dimensional) array stores values in a single sequence, like one line of data.


```python
cost = np.array([1000, 2000, 3000, 4000])
print(cost)
```

Create a 2D array

A 2D (two-dimensional) array stores values in rows and columns, like an Excel table.

```python
project_data = np.array([
    [1000,5],
    [2000,10],
    [3000, 15]
])

print(project_data)
```

```python
data = np.array([[
    [1000, 2],
    [2000, 3],
    [3000, 4]
]])

print(data)
```

Indexing — Select a specific value

Meaning: Indexing is used to access one particular element from an array.

```python
cost = np.array([1000,2000,3000,4000,5000])

print(cost[0])
print(cost[1])
print(cost[-1])
```

2. Slicing — Select multiple values.

Meaning: Slicing extracts a range of values from an array.

syntax : array[start:stop]

```python
print(cost[1:4])
print(cost[:3])
print(cost[2:])
```

3. Filtering — Select values based on a condition

Meaning: Filtering returns only the values that satisfy a condition.

```python
high_cost = cost[cost > 2500]
print(high_cost)

print(cost[cost < 4000])
print(cost[cost == 3000])
```

4. Reshaping — Change the array's structure

Meaning: Reshaping changes the number of rows and columns without changing the number of elements.

```python
numbers = np.array([1, 2, 3, 4, 5, 6])

new_array = numbers.reshape(2, 3)

print(new_array)

print(new_array.shape)
```

DAY- 7  Pandas II: Data Cleaning 

Task: Learn data cleaning and preprocessing techniques

```python
import pandas as pd

sales_df = pd.read_excel(
    r"E:\INTERNSHIP TASK\pandas_day6_sales_practice.xlsx",
    sheet_name="SalesData"
)


# 1 - Identify and handle missing values.  
print("\nmissing values: ")
print(sales_df.isnull().sum())

print(type(sales_df))

#  Fill missing SalesAmount using Quantity * UnitPrice
sales_df["SalesAmount"] = sales_df["SalesAmount"].fillna(sales_df["Quantity"] * sales_df["UnitPrice"])

print("Missing values after cleaning:")
print(sales_df.isnull().sum())
```

```python
#2 - Detect and remove duplicate records when appropriate.  
# Count duplicate rows
print("Duplicate Rows:")
print(sales_df.duplicated().sum())

# display duplicated rows
Duplicated_rows = sales_df[sales_df.duplicated()]
print(Duplicated_rows)

# REMOVE duplicates
sales_df = sales_df.drop_duplicates()

print("Duplicate rows after cleaning:", sales_df.duplicated().sum())
```

```python
#3 - Correct data types.
print(sales_df.dtypes)

# convert orderdate to datetime 
sales_df["OrderDate"] = pd.to_datetime(sales_df["OrderDate"], errors='coerce')

print(sales_df["OrderDate"].dtypes)
```

```python
# Standardize inconsistent text categories
print("Region:",  sales_df["Region"].unique())

print("Product:", sales_df["Product"].unique())

print("Category:", sales_df["Category"].unique())

print("PaymentMethod: ", sales_df["PaymentMethod"].unique())

print("OrderStatus: ", sales_df["OrderStatus"].unique())


# Step 2: Standardize text values
text_columns = [
    "Region",
    "Product",
    "Category",
    "PaymentMethod",
    "OrderStatus"
]

for col in text_columns:
    sales_df[col] = sales_df[col].str.strip().str.title()

print("Text standardization completed!")


# .str.strip() removes unwanted spaces at the beginning and end.
# .str.title() standardizes capitalization. For example, COMPLETED becomes Completed.
```

```python
# 5 - Convert date columns into proper datetime values.  
print(sales_df["OrderDate"].dtypes)

print(sales_df[["OrderID", "OrderDate"]].head())
print("Missing or invalid dates:", sales_df["OrderDate"].isna().sum())
```

DAY 10 — Statistics for Data Analytics and Data Science 

Task: Learn descriptive statistics and basic statistical interpretation. 

```python
#  Descriptive Statistics
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file_path = r"E:\INTERNSHIP TASK\pandas_day6_sales_practice.xlsx"

df = pd.read_excel(file_path)

print(df.head())
print(df.columns)


# Calculate Mean and Median
df[["Quantity", "UnitPrice", "SalesAmount"]].agg(
    ["mean", "median"]
)
```

```python
# Calculate Mode
df[["Quantity", "UnitPrice", "SalesAmount"]].mode()
```

```python
# Calculate Variance
df[["Quantity", "UnitPrice", "SalesAmount"]].var()

# Variance: Measures how spread out the values are from the mean, using squared deviations.
```

```python
# Calculate Standard deviation
df[["Quantity", "UnitPrice", "SalesAmount"]].std()

# Standard deviation: Measures spread in the original units, such as rupees for SalesAmount.
```

```python
# Calculate Quartiles
df[["Quantity", "UnitPrice", "SalesAmount"]].quantile(
    [0.25, 0.50, 0.75]
)

# Q1: 25% of the non-missing values are at or below Q1.
# Q2: 50% of the non-missing values are at or below Q2. It is the median.
# Q3: 75% of the non-missing values are at or below Q3.
```

```python

# Outlier Detection using IQR

sales = df["SalesAmount"].dropna()

Q1 = sales.quantile(0.25)
Q3 = sales.quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = sales[(sales < lower_limit) | (sales > upper_limit)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:")
print(outliers)
```

I used the IQR method to detect outliers in the SalesAmount column. The upper limit was ₹106,812.50. One sales value, ₹110,000, was above this limit and was identified as an outlier. This value should be investigated, but it should not be removed automatically.

Histogram: Shows how frequently sales amounts fall within different ranges.

Box plot: Shows the median, quartiles, spread, and potential outliers.

```python

# Histogram of Sales Amount

sales = df["SalesAmount"].dropna()

plt.figure(figsize=(8, 5))
plt.hist(sales, bins=10, edgecolor="black")
plt.title("Distribution of Sales Amount")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.show()
```

```python

# Box Plot of Sales Amount

plt.figure(figsize=(8, 4))
plt.boxplot(sales)
plt.title("Box Plot of Sales Amount")
plt.ylabel("Sales Amount")
plt.show()
```

Histogram: Displays the frequency distribution of sales amounts.

Box plot: Displays the median, quartiles, spread, and potential outliers.

Observation: The IQR method identified ₹110,000 as a potential outlier because it exceeds the upper limit of ₹106,812.50.

```python
# Calculate Correlation

correlation = df[
    ["Quantity", "UnitPrice", "SalesAmount"]
].corr()

print(correlation)
```

Correlation measures the strength and direction of a linear relationship between two numerical variables. In this dataset, UnitPrice and SalesAmount have a strong positive correlation of 0.779. Quantity and UnitPrice have a moderate negative correlation of −0.567. Quantity and SalesAmount have a weak negative correlation of −0.146.

```python

# Compare Mean and Median

sales = df["SalesAmount"].dropna()

print("Mean Sales Amount:", sales.mean())
print("Median Sales Amount:", sales.median())
```

The mean SalesAmount is ₹37,750, while the median is ₹24,500. The mean is higher than the median because larger sales values, including the ₹110,000 outlier, pull the average upward. The median is less affected by extreme values and may better represent a typical order in this dataset.

DAY 11 — Data Visualization with Matplotlib and Seaborn 

Task: Learn to visualize data and communicate findings. 

```python
# Step 1 — Import the libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Step 2 — Load your sales dataset
df = pd.read_excel(
    r"E:\INTERNSHIP TASK\pandas_day6_sales_practice.xlsx"
)

df.head()


print(df.columns.tolist())
print(df.shape)
df.info()
```

```python
# Chart 1: Bar chart — Sales by category
sales_by_category = df.groupby("Category")["SalesAmount"].sum()

sales_by_category.plot(kind="bar",figsize=(8,5))

plt.title("Total sales by category")
plt.xlabel("Categoty")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Line Chart
df["OrderDate"] = pd.to_datetime(df["OrderDate"])

monthly_sales = (
    df.set_index("OrderDate")["SalesAmount"]
      .resample("ME")
      .sum()
)

monthly_sales.plot(kind="line", marker="o", figsize=(9, 5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.legend(["Sales"])
plt.tight_layout()
plt.show()
```

```python
#Chart 3: Histogram
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="SalesAmount", bins=10, kde=True)

plt.title("Distribution of Sales Amounts")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Chart 4: Box Plot
# Sales outlier analysis
plt.figure(figsize=(8, 4))

sns.boxplot(data=df, x="SalesAmount")

plt.title("Sales Amount Outlier Analysis")
plt.xlabel("Sales Amount")
plt.tight_layout()
plt.show()
```

```python
# Chart 5: Scatter Plot
# Quantity vs Sales Amount
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Quantity",
    y="SalesAmount"
)

plt.title("Quantity vs Sales Amount")
plt.xlabel("Quantity")
plt.ylabel("Sales Amount")
plt.tight_layout()
plt.show()
```

```python
# Chart 6: Correlation Heatmap
numeric_df = df[
    ["Quantity", "UnitPrice", "SalesAmount"]
]

correlation = numeric_df.corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
